```python
from langchain_community.document_loaders import PyPDFLoader

# 1. Unga PDF file name-ah inga podunga
loader = PyPDFLoader("sample.pdf")
documents = loader.load()

# 2. Total pages evlo irukku nu check panrathu
print(f"Total pages loaded: {len(documents)}")
```

```python
!pip install langchain-text-splitters
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

# 1. Chunk size matrum overlap-ah set panrathu
text_splitter = RecursiveCharacterTextSplitter(chunk_size=500, chunk_overlap=50)

# 2. Document-ah chunks-ah cut panrathu
chunks = text_splitter.split_documents(documents)

# 3. Total ah evlo chunks vanthurukku nu check panrathu
print(f"Total chunks created: {len(chunks)}")
```

```python
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import FAISS

# 1. Free HuggingFace embeddings model-ah load panrom
embeddings = HuggingFaceEmbeddings(model_name="all-MiniLM-L6-v2")

# 2. Chunks-ah vector database (FAISS)-kku mathi store panrom
vector_db = FAISS.from_documents(chunks, embeddings)

print("Vector Database successfully created and ready!")
```

```python
# 1. User question-ah inga type pannikonga
query = "What is this document about?"

# 2. Vector database-la irunthu antha question-kku matching-ana docs-ah thedi edukkurom
relevant_docs = vector_db.similarity_search(query, k=2)

# 3. Output-ah print panrom
print("--- User Question ---")
print(query)
print("\n--- Bot Found Relevant Context from PDF ---")
for i, doc in enumerate(relevant_docs):
    print(f"\nResult {i+1}:")
    print(doc.page_content)
```
